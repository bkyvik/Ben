```python
print(df["latitude"].median())
print(df["longitude"].median())
```

```python
df["latitude_distance"] = (df["latitude"] - df["latitude"].median()).abs()
print(df)
```

```python
df["longitude_distance"] = (df["longitude"] - df["longitude"].median()).abs()
print(df)
```

```python
print(df.shape)
```

```python
min_lat_dist = df["latitude_distance"].min()
min_lat_dist_state = df["name"][df["latitude_distance"].idxmin()]
print(f"{min_lat_dist_state} is the closest to the median latitude, only {min_lat_dist:.2f} away")
```

```python
max_lat_dist = df["latitude_distance"].max()
max_lat_dist_state = df["name"][df["latitude_distance"].idxmax()]
print(f"{max_lat_dist_state} is the farthest to the median latitude, {max_lat_dist:.2f} away")
```

```python
min_long_dist = df["longitude_distance"].min()
min_long_dist_state = df["name"][df["longitude_distance"].idxmin()]
print(f"{min_long_dist_state} is the closest to the median longitude, only {min_long_dist:.2f} away")
```

```python
max_long_dist = df["longitude_distance"].max()
max_long_dist_state = df["name"][df["longitude_distance"].idxmax()]
print(f"{max_long_dist_state} is the farthest to the median longitude, {max_long_dist:.2f} away")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print(sns.get_dataset_names())
df2 = sns.load_dataset("car_crashes")
```

```python
print(df2)
```

```python
clean_df2 = df2.dropna(axis=1)
print(clean_df2.head())
```

```python
print(clean_df2.shape)
```

```python
# The shape is (51, 8)
# There are 51 rows and 8 columns
```

```python
print(clean_df2.head())
print(clean_df2.tail())
```

```python
# Column names:
# total = total car crash rate
# speeding = proportion of crashes involving speeding
# alcohol = proportion of crashes involving alcohol
# not_distracted = proportion of crashes wehre driver was not distracted
# no_previous = average number of driver's previous crashes
# ins_premium = average insurance premium for crash
# ins_losses = average insurance losses for crash
# abbrev = state where crahses are analyzed
```

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].scatter(clean_df2["total"], clean_df2["speeding"])
axes[0, 0].set_title("Total crashes vs. speeding crashes")
axes[0, 0].set_xlabel("Total crashes")
axes[0, 0].set_ylabel("Speeding crashes")

axes[0, 1].scatter(clean_df2["total"], clean_df2["alcohol"], color = "red", label = "State's emalgamated crashes")
axes[0, 1].set_title("Total crashes vs. drunk crashes")
axes[0, 1].set_xlabel("Total crashes (per 100K people)")
axes[0, 1].set_ylabel("Drunk crashes (per crash)")
axes[0, 1].legend()

sns.histplot(clean_df2["ins_premium"], ax=axes[1, 0], color = "yellow")
axes[1, 0].set_title("Insurance premium distribution")
axes[1, 0].set_xlabel("Insurance premium (dollars)")
axes[1, 0].set_ylabel("Count")

sns.boxplot(clean_df2["ins_losses"], ax=axes[1, 1], color = "green")
axes[1, 1].set_title("Insurance losses distribution")
axes[1, 1].set_ylabel("Dollars")

plt.tight_layout()
plt.show()
```

```python
# The first one compares the rate of overal crashes to
# the proportion of speed crashes, showing the effect
# speeding on overal crash rate
# I learned that the two values have a positive correlation

# The second compares rate of overall crashes to proportion
# of alcohol-related crashes, showing the effect of alcohol
# on overall crash rate
# I learned that the two values have a positive correlation

# The third shows the distribution of average insurance
# premiums across each state, showing how many states
# have various average premiums
# I learned that the average insurance premium tends to
# be between $800 and $1000

# The fourth evaluates the range and median distributions
# of average insurance losses across each state, showing
# a national average insurance loss
# I learned that the median insurance loss from car crashes is ~$130
```
