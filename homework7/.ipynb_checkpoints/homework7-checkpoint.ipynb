```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axs = plt.subplots(1, 3, figsize=(25, 5))
for A in amplitude:
    axs[0].plot(x, make_sine_wave(x, A, 1), label=f"A = {A}")
    axs[0].set_title("Sine with varying amplitudes")
    axs[0].set_xlabel("X values")
    axs[0].set_ylabel("Y values")
    axs[0].legend()
for w in frequency:
    axs[1].plot(x, make_sine_wave(x, 1, w), label=f"w = {w}")
    axs[1].set_title("Sine with varying frequencies")
    axs[1].set_xlabel("X values")
    axs[1].set_ylabel("Y values")
    axs[1].legend()
for A, w in zip(amplitude, frequency):
    axs[2].plot(x, make_sine_wave(x, A, w), label="A = {A}, w = {w}")
    axs[2].set_title("Sine with varying amplitudes and frequencies")
    axs[2].set_xlabel("X values")
    axs[2].set_ylabel("Y values")
    axs[2].legend()
plt.show()
```
