```python
# 3.1 Importing Data with NumPy
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows = 1)
```

```python
# 3.2 Analyzing Imported Data
print(data.shape)
# Assuming I did this right, there are 2642 rows
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
min_magn = min(magnitude)
min_magn_star = np.argmin(magnitude)
print(f"The star with the lowest magnitude is star #{min_magn_star} with magnitude: {min_magn}")
max_magn = max(magnitude)
max_magn_star = np.argmax(magnitude)
print(f"The star with the highest magnitude is star #{max_magn_star} with magnitude: {max_magn}")
```

```python
min_temp = min(temperature)
min_temp_star = np.argmin(temperature)
print(f"The star with the lowest temperature is star #{min_temp_star} with temperature: {min_temp}")
max_temp = max(temperature)
max_temp_star = np.argmax(temperature)
print(f"The star with the highest temperature is star #{max_temp_star} with tempurature: {max_temp}")
```

```python
print(f"The brightest star is #{min_magn_star}, and the dimmest is #{max_magn_star}")
```

```python
# 3.3 Plot and HR Diagram
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, label = "star")
plt.scatter(log_temperature[min_magn_star], log_magnitude[min_magn_star], c = "black", marker = "*", s = 300, label = "brightest")
plt.scatter(log_temperature[max_magn_star], log_magnitude[max_magn_star], c = "purple", marker = "*", s = 300, label = "dimmest")
plt.scatter(log_temperature[max_temp_star], log_magnitude[max_temp_star], c = "green", marker = "*", s = 300, label = "hottest")
plt.scatter(log_temperature[min_temp_star], log_magnitude[min_temp_star], c = "orange", marker = "*", s = 300, label = "coldest")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```
